# Music Recommender System — Serving Walkthrough

This notebook demonstrates the API endpoints and CLI commands
for getting artist and track recommendations.

## Prerequisites

1. Train the model: `uv run music-recommender train`
2. Start the API: `uvicorn api.main:app --reload`

The API runs on `http://localhost:8000` by default.

````python
import requests

BASE_URL = "http://localhost:8000"

# Health check
resp = requests.get(f"{BASE_URL}/health")
print(f"Health: {resp.json()}")
````

## Artist recommendations

````python
resp = requests.get(
    f"{BASE_URL}/recommend/user/user_1",
    params={"top_k": 5, "explain": True},
)
print(f"Status: {resp.status_code}")
data = resp.json()
print(f"Strategy: {data.get('strategy', 'N/A')}")
for rec in data.get("recommendations", []):
    reasons = "; ".join(rec.get("reasons", []))
    print(f"  {rec['artist_name']} ({rec['artist_id']}) - score: {rec['score']:.4f} | {reasons}")
````

## Track recommendations

````python
resp = requests.get(
    f"{BASE_URL}/recommend/tracks/user_1",
    params={"top_k": 5, "explain": True},
)
print(f"Status: {resp.status_code}")
data = resp.json()
for rec in data.get("recommendations", []):
    reasons = "; ".join(rec.get("reasons", []))
    print(f"  {rec['track_name']} by {rec.get('artist_name', '?')} (score: {rec['score']:.4f}) | {reasons}")
````

## CLI alternative

The same recommendations are available from the CLI without starting
the API:

```bash
# Artist recommendations
uv run music-recommender recommend-user --user-id user_1 --top-k 5 --explain

# Track recommendations
uv run music-recommender track-recommendations --user-id user_1 --top-k 5 --explain
```
