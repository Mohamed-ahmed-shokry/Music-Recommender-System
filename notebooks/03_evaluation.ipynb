# Music Recommender System — Evaluation Walkthrough

This notebook demonstrates how to evaluate track recommendations
using holdout evaluation with multiple strategies and metrics.

````python
from music_recommender.data import (
    load_and_validate_track_interactions,
    load_and_validate_track_metadata,
)
from music_recommender.track_evaluate import evaluate_track_holdout

df = load_and_validate_track_interactions("data/raw/sample_track_interactions.csv")
meta = load_and_validate_track_metadata("data/raw/sample_track_metadata.csv", df)
print(f"Loaded {df.shape[0]} interactions, {meta.shape[0]} tracks")
````

## Single-strategy evaluation

Run a basic holdout evaluation with the similarity strategy.

````python
metrics = evaluate_track_holdout(df, meta, top_k=5, folds=2)
for name, value in sorted(metrics.items()):
    print(f"  {name}: {value:.4f}")
````

## Compare all strategies

Use `compare_all=True` to run similarity, popularity, and hybrid
strategies in a single pass.

````python
comparison = evaluate_track_holdout(df, meta, top_k=5, folds=2, compare_all=True)
for strategy in ["similarity", "popularity", "hybrid"]:
    print(f"\n{strategy.upper()}:")
    for name, value in sorted(comparison[strategy].items()):
        print(f"  {name}: {value:.4f}")
````

## CLI equivalent

The same evaluation can be run from the CLI:

```bash
# Single strategy
uv run music-recommender evaluate-tracks --top-k 5 --folds 2

# Compare all strategies
uv run music-recommender evaluate-tracks --top-k 5 --folds 2 --compare-all

# Save a report
uv run music-recommender evaluate-tracks --top-k 5 --folds 2 --compare-all --report-name eval_run
```
