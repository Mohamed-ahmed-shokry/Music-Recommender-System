# Music Recommender System — Data Exploration

This notebook walks through the raw data and trained artifacts to
understand the system's inputs and outputs.

```python
import pandas as pd

interactions = pd.read_csv("data/raw/sample_interactions.csv")
artist_meta = pd.read_csv("data/raw/sample_artist_metadata.csv")
track_int = pd.read_csv("data/raw/sample_track_interactions.csv")
track_meta = pd.read_csv("data/raw/sample_track_metadata.csv")

n_users = interactions["user_id"].nunique()
n_artists = interactions["artist_id"].nunique()
print(
    f"Artist interactions: {len(interactions)} rows, "
    f"{n_users} users, {n_artists} artists"
)

tn_users = track_int["user_id"].nunique()
tn_tracks = track_int["track_id"].nunique()
print(
    f"Track interactions:  {len(track_int)} rows, {tn_users} users, {tn_tracks} tracks"
)
print(f"Artist metadata:    {len(artist_meta)} artists")
print(f"Track metadata:     {len(track_meta)} tracks, {track_meta.shape[1]} columns")
```

```python
interactions.head(10)
```

```python
audio_cols = [
    "track_name",
    "artist_name",
    "danceability",
    "energy",
    "valence",
    "tempo",
]
track_meta[audio_cols].head(10)
```

## Play-count distribution

A few users dominate play counts — this is typical for music data
and drives the popularity bias that the ranking knobs address.

```python
user_plays = (
    interactions.groupby("user_id")["play_count"].sum().sort_values(ascending=False)
)
print(user_plays.describe())
user_plays
```

## Audio-feature correlation

The 12 Spotify-style features used for cosine-similarity content
recommendations.

```python
feat_cols = [
    "danceability",
    "energy",
    "speechiness",
    "acousticness",
    "instrumentalness",
    "liveness",
    "valence",
]
track_meta[feat_cols].describe().round(3)
```
