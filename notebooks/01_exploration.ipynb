# Music Recommender System — Data Exploration

This notebook walks through the raw data and trained artifacts to
understand the system's inputs and outputs.

```python
import pandas as pd

interactions = pd.read_csv("data/raw/sample_interactions.csv")
artist_meta = pd.read_csv("data/raw/sample_artist_metadata.csv")
track_interactions = pd.read_csv("data/raw/sample_track_interactions.csv")
track_meta = pd.read_csv("data/raw/sample_track_metadata.csv")

print(f"Artist interactions: {interactions.shape[0]} rows, {interactions['user_id'].nunique()} users, {interactions['artist_id'].nunique()} artists")
print(f"Track interactions:  {track_interactions.shape[0]} rows, {track_interactions['user_id'].nunique()} users, {track_interactions['track_id'].nunique()} tracks")
print(f"Artist metadata:    {artist_meta.shape[0]} artists")
print(f"Track metadata:     {track_meta.shape[0]} tracks, {track_meta.shape[1]} columns")
```

## Play-count distribution

A few users dominate play counts — this is typical for music data and
drives the popularity bias that the ranking knobs address.

```python
user_plays = interactions.groupby("user_id")["play_count"].sum().sort_values(ascending=False)
print(user_plays.describe())
user_plays
```

## Audio-feature correlation

The 12 Spotify-style features used for cosine-similarity content
recommendations.

```python
audio_cols = ["danceability", "energy", "speechiness", "acousticness",
              "instrumentalness", "liveness", "valence"]
track_meta[audio_cols].describe().round(3)
```

Run `uv run music-recommender train` from the repo root to train
the ALS model and build the artifact bundle. The command saves
model, mappings, and the `recommender_artifact.pkl` to `artifacts/`.
